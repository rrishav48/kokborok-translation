```python
import pandas as pd
import re

# Load the dataset
data = pd.read_csv('kokdata2copy.csv')

# Normalize the dataset to lowercase and strip spaces
data['English'] = data['English'].astype(str).str.lower().str.strip()
data['Kokborok'] = data['Kokborok'].astype(str).str.strip()

# Create dictionary for lookup
translation_dict = dict(zip(data['English'], data['Kokborok']))

# Clean word by removing punctuation for lookup
def clean_word(word):
    return re.sub(r'[^\w\s]', '', word).lower()

def translate_to_kokborok(sentence):
    # Extract words + punctuation separately
    tokens = re.findall(r"\w+|[^\w\s]", sentence, re.UNICODE)

    translated_sentence = []
    i = 0

    while i < len(tokens):
        word = clean_word(tokens[i])
        next_word = clean_word(tokens[i + 1]) if i + 1 < len(tokens) else ""

        # Attempt two-word phrase
        if next_word:
            phrase = f"{word} {next_word}"
            if phrase in translation_dict:
                translated = translation_dict[phrase]
                translated_sentence.append(translated)
                i += 2
                continue

        # Single word translation
        if word in translation_dict:
            translated = translation_dict[word]
        elif tokens[i].isdigit():
            translated = tokens[i]  # Keep numbers as-is
        elif re.match(r'[^\w\s]', tokens[i]):
            translated = tokens[i]  # Keep punctuation
        else:
            translated = f"[{tokens[i]}]"  # mark as untranslated

        translated_sentence.append(translated)
        i += 1

    return ' '.join(translated_sentence)

# Example usage
if __name__ == "__main__":
    print("Type 'exit' to quit the translator.")
    while True:
        user_input = input("\nEnter an English sentence: ")
        if user_input.lower() == "exit":
            print("Goodbye! 👋")
            break
        kokborok_translation = translate_to_kokborok(user_input)
        print(f"Kokborok translation: {kokborok_translation}")



```
